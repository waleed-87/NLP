```python

import requests
import time

def fetch_all_api_data():
    all_frames = []


    for i in range(1, 472):

        url = f"https://api.themoviedb.org/3/movie/top_rated?api_key=8265bd1679663a7ea12ac168da84d2e8&language=en-US&page={i}"

        try:
            response = requests.get(url)

            if response.status_code == 200:
                data = response.json()


                page_df = pd.DataFrame(data)


                if not page_df.empty:
                    all_frames.append(page_df)
                    print(f"Successfully fetched page {i}/{471}")
            else:
                print(f"Skipped page {i}: Received status code {response.status_code}")

        except Exception as e:
            print(f"Error on page {i}: {e}")


        time.sleep(0.1)


    if all_frames:
        final_df = pd.concat(all_frames, ignore_index=True)
        return final_df
    else:
        print("No data was collected.")
        return pd.DataFrame()


df = fetch_all_api_data()
print(f"Done! Total rows collected: {len(df)}")
```

```python
import re
def remove_html_tags(text):
    pattern = re.compile('<.*?>')
    return pattern.sub(r'', text)
```

```python
df['overview'] = df['overview'].apply(remove_html_tags)
```

```python
def remove_url(text):
    pattern = re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub(r'', text)
```

```python
df['overview'] = df['overview'].apply(remove_url)
```

```python
# 1. Install symspellpy
!pip install symspellpy

import pkg_resources
from symspellpy import SymSpell, Verbosity

# 2. Initialize and load the dictionary (takes a moment)
sym_spell = SymSpell(max_dictionary_edit_distance=2, prefix_length=7)
dictionary_path = pkg_resources.resource_filename(
    "symspellpy", "frequency_dictionary_en_82_765.txt"
)
sym_spell.load_dictionary(dictionary_path, term_index=0, count_index=1)

# 3. Fast correction function
def fast_correct(text):
    # lookup_compound handles entire sentences/paragraphs at once
    suggestions = sym_spell.lookup_compound(text, max_edit_distance=2)
    return suggestions[0].term if suggestions else text

# 4. Apply it instantly using Option A's list structure
overviews = df['overview'].astype(str).tolist()
df['corrected_overview'] = [fast_correct(text) for text in overviews]
```

```python
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')

# OPTIMIZATION: Convert the list to a SET. Lookups in a set take constant time O(1).
stop_words_set = set(stopwords.words('english'))

def stop_word(txt):
    # Split text into words, filter out stop words, and skip empty elements
    cleaned_words = [word for word in txt.split() if word.lower() not in stop_words_set]
    return " ".join(cleaned_words)

# Apply it instantly using the fast list approach we used before
overviews = df['overview'].astype(str).tolist()
df['cleaned_overview'] = [stop_word(text) for text in overviews]
```

```python
import spacy
nlp = spacy.load('en_core_web_sm')
```

```python
cleaned_docs = []
for doc in nlp.pipe(df['cleaned_overview'].astype(str), batch_size=500, disable=["parser", "ner"]):
    # Extract whatever you need (e.g., lowercase words, or lemmatized words)
    tokens = [token.text for token in doc]
    cleaned_docs.append(tokens)

# 3. Assign it back to the DataFrame
df['tokenized_overview'] = cleaned_docs
```

```python
import nltk
from nltk.stem import WordNetLemmatizer

# Ensure the WordNet dictionary is downloaded
nltk.download('wordnet')
wordnet_lemmatizer = WordNetLemmatizer()

# 1. Define your punctuation set
punctuations = "?:!.,;"

# 2. Use .apply() to clean and lemmatize row-by-row
# Note: This assumes df['tokenized_overview'] is already a list of words from your spaCy code
df['lemmatized_overview'] = df['tokenized_overview'].apply(
    lambda tokens: [
        wordnet_lemmatizer.lemmatize(word)  # Lemmatize the word
        for word in tokens                  # Loop through each word in the row
        if word not in punctuations         # Strip out punctuation
    ]
)
```
